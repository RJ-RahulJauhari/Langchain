# Chat Models — OpenAI (GPT-4o)

## What is a Chat Model?

A **Chat Model** takes a **list of messages** as input (not just a single string) and returns a response message. This design mimics a real conversation and lets the model understand context from previous turns.

Each message has a **role**:
- `SystemMessage` — Background instructions for the AI. The user never sees this — it shapes the AI's personality and behaviour. Example: *"You are a helpful doctor who explains things simply."*
- `HumanMessage` — What the user types.
- `AIMessage` — What the AI replied (used to give the model memory of past turns).

## Why this design?

By passing the full conversation history as a list, the model can understand context like:
- "What did I mean by 'it' earlier?"
- "You mentioned Python in your last message — I'll continue on that topic."

The model has no built-in memory — you pass the history every time you call `.invoke()`.

## What you'll learn in this notebook

- How to create a `ChatOpenAI` model with custom settings
- How to invoke it with a message and read the response
- What the full response object contains (not just the text)
- How to stream responses word-by-word

## Prerequisites

- `OPENAI_API_KEY` in `.env`
- Virtual environment activated

```python
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
load_dotenv()
```

```python
chat_model = ChatOpenAI(model="gpt-4o", temperature=1.5, max_completion_tokens=200)
```

```python
response = chat_model.invoke("What is the best way to learn Langchain?")
response
```

```python
response.content
```

```python
# --- Section: Streaming Responses ---
# Instead of waiting for the full response, .stream() yields it chunk by chunk.
# This makes your app feel faster and more responsive — you see words appear
# as they're generated, just like ChatGPT's typing effect.

from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage

streaming_model = ChatOpenAI(model="gpt-4o")

print("Streaming response:")
for chunk in streaming_model.stream([HumanMessage("Write a haiku about Python programming.")]):
    print(chunk.content, end="", flush=True)
print()  # newline after streaming ends
```

## What you learned in this notebook

| Concept | Code |

| Single message call | `model.invoke("message")` |
