# Chat Models — Ollama (Local Models)

## What is Ollama?

**Ollama** is a tool that lets you run large language models **completely on your own machine** — no internet required, no API key, no cost per request, total privacy.

You download a model once (e.g. `llama3.1`, `qwen3`, `gemma`), and Ollama runs it as a local server. LangChain's `ChatOllama` talks to this server exactly like it would talk to OpenAI — the code looks almost identical.

## Why use local models?

| Cloud (e.g. OpenAI) | Local (e.g. Ollama) |
|--------------------|---------------------|
| Best quality models | Good quality, free |
| Costs money per token | Free after download |
| Your data leaves your machine | 100% private |
| Always available | Needs enough RAM/VRAM |

For learning and development, Ollama is perfect. For production with high quality requirements, cloud models are usually better.

## What you'll learn in this notebook

- How to use `ChatOllama` with a system message to give the model a personality
- How to pass a conversation history to the model
- How the model's response includes metadata (tokens, timing)
- How to swap between OpenAI and Ollama with almost no code change

## Prerequisites

- Ollama installed and running (`ollama serve`)
- A model pulled — e.g. `ollama pull llama3.1` (smaller) or `ollama pull qwen3-coder:30b` (larger, slower)

```python
from langchain_ollama import ChatOllama
from langchain_core.messages import HumanMessage, SystemMessage
```

```python
opensource_chat_model = ChatOllama(model="qwen3-coder:30b", temperature=1.5)
```

```python
system_message = SystemMessage("You are a sassy helpful agent, please answer all queries in a sassy.")
```

```python
history = [system_message,HumanMessage("What is you latest context, till what date you are trained?")]
```

```python
response = opensource_chat_model.invoke(history)
```
