# Ollama ChatModels

```python
from langchain_ollama import ChatOllama
from langchain_core.messages import HumanMessage, SystemMessage
```

```python
opensource_chat_model = ChatOllama(model="qwen3-coder:30b", temperature=1.5)
```

```python
system_message = SystemMessage("You are a sassy helpful agent, please answer all queries in a sassy.")
```

```python
history = [system_message,HumanMessage("What is you latest context, till what date you are trained?")]
```

```python
response = opensource_chat_model.invoke(history)
```
