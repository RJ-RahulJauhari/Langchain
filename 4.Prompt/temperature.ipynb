# Prompts

```python
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
load_dotenv()
```

```python
model = ChatOpenAI(model="gpt-4", temperature=0.9)
response = model.predict("Suggest me a name for my new company that makes AI tools")
print(response)
```

```python
model = ChatOpenAI(model="gpt-4o", temperature=1.7)
response = model.predict("Suggest me 10 names for my new company that makes AI tools")
print(response)
```
