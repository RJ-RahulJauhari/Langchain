# Simple Chatbot — Conversation Memory with a List

## How does a chatbot "remember" past messages?

LLMs have **no built-in memory**. Every time you call `.invoke()`, the model starts completely fresh — it has no idea what you said 2 messages ago.

The solution is simple: **you keep a list of all messages and pass the entire list to the model every single time**.

```
Turn 1: model.invoke([system, human("Hi!")])
Turn 2: model.invoke([system, human("Hi!"), ai("Hello!"), human("What's my name?")])
Turn 3: model.invoke([system, human("Hi!"), ai("Hello!"), human("What's my name?"), ai("I don't know"), human("My name is Rahul")])
```

The list IS the memory. The longer the conversation, the more tokens you send each time. That's the trade-off.

## What this notebook builds

A simple infinite loop chatbot that:
1. Shows a prompt: `"You: "`
2. Adds your message to `chat_history`
3. Calls the model with the full history
4. Prints the response
5. Adds the response to `chat_history`
6. Repeats until you type `exit`

## What you'll learn

- How conversation history is implemented in practice
- How `HumanMessage` and `AIMessage` build up the context list
- How `SystemMessage` sets the AI's personality at the start
- Why the model needs the full history on every call

````python
from langchain_ollama import ChatOllama
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage
from langchain_core.prompts import PromptTemplate
````

````python
print("Welcome to the Chatbot!")
````

````python
MODEL = "qwen3-coder:30b"
TEMPERTURE = 1.5
````

````python
chat_history = [SystemMessage("Your are the best and greatest programmer in the world, you will write production ready code, with all the edge cases handled. You will also write tests for your code. You will also explain your code. You will also write documentation for your code. You will also write comments for your code. You will also write type hints for your code. You will also write code that is easy to read and understand. You will also write code that is easy to maintain. You will also write code that is easy to extend. You will also write code that is easy to debug. You will also write code that is easy to test. You will also write code that is easy to deploy. You will also write code that is easy to scale. You will also write code that is easy to monitor. You will also write code that is easy to log. You will also write code that is easy to secure. You will also write code that is easy to optimize. You will also write code that is easy to refactor. You will also write code that is easy to document. You will also write code that is easy to comment. You will also write code that is easy to type hint."),
                HumanMessage(content="Hello, how are you?")
                ]
````

````python
chat_model = ChatOllama(model=MODEL, temperature=TEMPERTURE)
````

````python
while True:
    user_input = input("You: ")
    if user_input.lower() in ["exit", "quit"]:
        print("Exiting the chatbot. Goodbye!")
        break

    chat_history.append(HumanMessage(content=user_input))
    response = chat_model.invoke(chat_history)
    print(f"Bot: {response.content}")
    chat_history.append(AIMessage(content=response.content))
````
