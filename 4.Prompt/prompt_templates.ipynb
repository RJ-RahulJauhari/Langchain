# Prompt Templates — Reusable, Parameterised Prompts

## What is a Prompt Template?

A **prompt template** is a reusable prompt with **placeholders** (variables) you fill in at runtime.

**Without a template** (hardcoded — not reusable):
```python
model.invoke("You are a cricket expert. Explain Virat Kohli in simple terms.")
model.invoke("You are a cricket expert. Explain LBW rule in simple terms.")
model.invoke("You are a football expert. Explain Messi in simple terms.")
```

**With a template** (one template, infinite uses):
```python
template = "You are a {domain} expert. Explain {topic} in simple terms."
# Call with any domain + topic combination
```

Templates make your code:
- **Reusable** — one definition, hundreds of uses
- **Readable** — the structure is visible at a glance
- **Testable** — you can inspect the filled prompt before sending it

## PromptTemplate vs. ChatPromptTemplate

| | PromptTemplate | ChatPromptTemplate |
|-|----------------|-------------------|
| **Output** | A single string | A list of messages |
| **Used with** | Plain LLMs | Chat models |
| **Structure** | `template = "Explain {topic}"` | `[("system", "..."), ("human", "...")]` |

## MessagesPlaceholder

`MessagesPlaceholder` is a special slot inside a `ChatPromptTemplate` where you can inject a dynamic list of past messages (conversation history). This is essential for chatbots that need to remember previous turns.

PromptTemplate is used for single turn messages and ChatPromptTemplate is for multi-turn conversations

````python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage
````

````python
chat_template = ChatPromptTemplate(
    [
        ("system", "You are a helpful {domain} expert..."),
        ("human","Explain in simple terms, what is {topic}")
    ]
)
````

````python
prompt = chat_template.invoke({"domain":"criket","topic":"Virat Kohli"})
````
