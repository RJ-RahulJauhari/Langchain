```python
from langchain_huggingface import HuggingFaceEmbeddings
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# HuggingFaceEmbeddings wraps any sentence-transformers model through LangChain's stable API.
# all-MiniLM-L6-v2 is a small (~90 MB), fast, and very widely used open-source model.
# It downloads automatically the first time and is cached locally on subsequent runs.
embedding_model = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")

query = "Which planet is known as the Red Planet?"
documents = [
    "Venus is often called Earth's twin because of its similar size and proximity.",
    "Mars, known for its reddish appearance, is often referred to as the Red Planet.",
    "Jupiter, the largest planet in our solar system, has a prominent red spot.",
    "Saturn, famous for its rings, is sometimes mistaken for the Red Planet.",
]

# Embed all documents into vectors
doc_embeddings = embedding_model.embed_documents(documents)
# Embed the query into the same vector space
query_embedding = embedding_model.embed_query(query)

print(f"Embedding dimensions: {len(doc_embeddings[0])}")

# Rank documents by cosine similarity to the query
scores = cosine_similarity([query_embedding], doc_embeddings)[0]
ranked = sorted(zip(scores, documents), reverse=True)

print(f"\nQuery: {query}\n")
print("Ranked by similarity (highest = most relevant):")
for score, doc in ranked:
    print(f"  {score:.4f} — {doc}")
```
