# Embedding Models — OpenAI

## What is a Text Embedding?

An **embedding** converts text into a list of numbers called a **vector**. The key property is: **text with similar meaning gets similar numbers**.

```
"The cat sat on the mat"   → [0.12, -0.43, 0.87, ...]  (384 or 1536 numbers)
"A feline rested on a rug" → [0.11, -0.41, 0.85, ...]  ← very similar numbers!
"The stock market crashed"  → [0.93, 0.21, -0.64, ...]  ← very different numbers
```

This is the foundation of:
- **Semantic search** — find relevant documents by meaning, not just keywords
- **RAG** (Retrieval-Augmented Generation) — give an LLM relevant context to answer questions
- **Recommendation systems** — suggest similar content

## What do the numbers mean?

Each number in the vector represents a "dimension" of meaning — like coordinates in a very high-dimensional space. The meaning of each individual number is not human-interpretable, but the relationship between vectors is.

If two vectors are "close together" in that space, the texts are similar in meaning. If they're "far apart", the texts are unrelated.

## OpenAI's embedding models

| Model | Dimensions | Notes |
|-------|-----------|-------|
| `text-embedding-3-small` | 1536 (default) | Fast, cheap, good quality |
| `text-embedding-3-large` | 3072 (default) | Best quality, supports custom dimensions |

You can reduce the `dimensions` parameter to save space at a slight quality cost.

## Prerequisites

- `OPENAI_API_KEY` in `.env`
- Virtual environment activated

````python
from langchain_openai import OpenAIEmbeddings
from dotenv import load_dotenv
````

````python
load_dotenv()
````

````python
embedding_model = OpenAIEmbeddings(model="text-embedding-3-large", dimensions=32)
````

````python
embedding = embedding_model.embed_query("Hello world")
````

````python
print(embedding)
````

````python
len(embedding)
````

````python
for i in embedding:
    print(i)
    print("\n")
````

````python
# --- Section: Understanding the Embedding Vector ---
# Let's look at what the vector actually represents.
# Higher absolute values indicate stronger activation in that "dimension" of meaning.
# The most important thing is NOT individual values, but the RELATIONSHIP between vectors.

print(f"Total dimensions: {len(embedding)}")
print(f"Min value: {min(embedding):.4f}")
print(f"Max value: {max(embedding):.4f}")
print(f"\nThe numbers themselves have no human-readable meaning.")
print(f"Their value is in how they COMPARE to other embeddings.")
print(f"\nFor example, similar texts will have vectors that point")
print(f"in the same direction — high cosine similarity.")
print(f"Unrelated texts will have vectors pointing different directions.")
print(f"\nSee notebook 4_Document_Similarity.ipynb for a live demonstration.")
````
