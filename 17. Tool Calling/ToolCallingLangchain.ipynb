# Tool Calling 

```python
from langchain_ollama import ChatOllama
from langchain_core.messages import HumanMessage
```

```python
model = ChatOllama(model="llama3.1:latest")
```

```python
model.invoke([HumanMessage(content="What is the meaning of life?")])    
```

```python
from langchain_core.tools import tool
from langchain_core.messages import HumanMessage
```

```python


@tool
def multiply(a: int, b: int) -> int:
    """Given 2 numbers this tool multiplies them together."""
    return a * b
```

```python
multiply.invoke({"a": 5, "b": 10})
```

```python
multiply.name
```

```python
multiply.description
```

```python
multiply.args
```

```python
model_with_tools = model.bind_tools([multiply])
```

```python
response = model_with_tools.invoke("What is 5 multiplied by 1 and 0?")
```
