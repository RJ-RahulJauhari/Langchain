```python
from langchain_ollama import ChatOllama
from langchain_community.tools import DuckDuckGoSearchRun
```

```python
model = ChatOllama(model="gemma4:26b")
```

```python
tool = DuckDuckGoSearchRun()
```

```python
tool.invoke("Top news in india today?")
```

```python
model.invoke("What is the capital of France?")
```

```python
from langchain.agents import create_agent
```

```python
tools = [DuckDuckGoSearchRun()]
```

```python
agent = create_agent(
    model=model,
    tools=tools,
    system_prompt="You are a helpful assistant, please use the tools to answer the question. If you don't know the answer, say you don't know.",
)
```

```python
agent.invoke({"input": "What is the capital of France?"})
```

```python
response = agent.invoke({"messages": [{"role": "user", "content": "What is the weather in Bangalore on 19th April 2026?"}]}
)["messages"][-1]
response
```
