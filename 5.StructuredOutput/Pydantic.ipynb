```python
class Review(BaseModel):
    key_themes: list[str] = Field(description="The key themes of the review")
    summary: str = Field(description="A brief summary of the review")
    sentiment: Literal["positive", "negative", "neutral"] = Field(description="The sentiment of the review, either positive, negative or neutral")
    pros: Optional[list[str]] = Field(description="The pros of the review")
    cons: Optional[list[str]] = Field(description="The cons of the review")
```

```python
from langchain_ollama import ChatOllama

model = ChatOllama(model="qwen3-coder:30b")
```

```python
review_model = model.with_structured_output(Review)
```

```python
review = ''' 
I’ve been using these wireless noise-cancelling headphones for a few weeks now, and overall they deliver a very solid experience for the price.

Pros
Excellent noise cancellation: Effectively blocks out office chatter and low-level background noise on public transport.
Comfortable fit: Soft ear cushions and adjustable headband make them easy to wear for 2–3 hours without discomfort.
Good sound quality: Clear mids and highs with a decent bass response; great for podcasts, calls, and most music genres.
Strong battery life: I get around 25–30 hours of playback on a single charge, even with noise cancellation turned on.
Stable Bluetooth connection: Pairs quickly with laptop and phone; minimal dropouts during calls or streaming.
Cons
Bulky design: The ear cups are a bit large, so they’re not the most portable option and take up space in a bag.
Average microphone quality: Call quality is acceptable but not great in windy or very noisy environments.
Limited customization: The companion app (if available) has only basic EQ options and no advanced sound profiles.
Plastic build: While lightweight, some parts feel a bit cheap and may not hold up to rough handling.

'''


review_model.invoke(review).model_dump_json()
```
