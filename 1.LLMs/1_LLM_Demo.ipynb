# LLM Demo — Your First Call to an AI Model

## What is an LLM?

**LLM** stands for **Large Language Model**. It's a type of AI that has been trained on enormous amounts of text (books, websites, code, articles) and learned to predict and generate human-like text.

When you send it a question or a piece of text (called a **prompt**), it generates a response based on everything it has learned. It doesn't "look up" answers — it generates them, word by word, based on patterns it learned during training.

**Famous LLMs:**
- GPT-3.5, GPT-4 (by OpenAI)
- Claude (by Anthropic)
- Llama, Gemma, Qwen (open-source, run locally)

## LLM vs. Chat Model

There are two "modes" for interacting with language models:

| Mode | Input | Best for |
|------|-------|---------|
| **LLM** (completion) | A single text string | "Complete this text…" |
| **Chat Model** | A list of messages (conversation) | "Back and forth dialogue" |

This notebook uses the **LLM** (completion) mode via OpenAI's `gpt-3.5-turbo-instruct` model. Chat models are covered in `2.ChatModels/`.

## Prerequisites

- `OPENAI_API_KEY` set in your `.env` file
- Virtual environment activated

```python
from langchain_openai import OpenAI
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
llm = OpenAI(model="gpt-3.5-turbo-instruct")
```

```python
llm.invoke("What is the capital of India?")
```

```python
print("Setup complete.")
```

## Batch Invocation

Instead of calling the model one prompt at a time, `.batch()` processes multiple prompts in a single call and returns a list of results. This is significantly faster when you have many prompts to handle.

```python
responses = llm.batch([
    "What is the capital of India?",
    "What is the capital of France?",
    "What is the capital of Japan?",
])

for i, resp in enumerate(responses, 1):
    print(f"Response {i}: {resp.strip()}")
```
